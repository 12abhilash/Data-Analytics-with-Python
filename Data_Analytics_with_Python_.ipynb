```python
#Frozenset
```

```python
a = frozenset({10, 20, 30, 40})

print(a)
print(type(a))
```

```python
#Boolean
```

```python
a = True
b = False

print(a)
print(b)

print(type(a))
```

```python
#Bytes
```

```python
a = bytes([65, 66, 67])

print(a)
print(type(a))
```

```python
#Bytearray
```

```python
a = bytearray([65, 66, 67])

a[0] = 68

print(a)
print(type(a))
```

```python
#Memoryview
```

```python
a = bytes([65, 66, 67])

b = memoryview(a)

print(b)
print(b[0])
print(type(b))
```

```python
#NoneType
```

```python
a = None

print(a)
print(type(a))
```
